```python
from datasets import load_dataset

ds = load_dataset("qiaojin/PubMedQA", "pqa_labeled")['train']
```

```python
extra_info = []

for i in range(len(ds)):
    context = "<context> " + " ".join(ds[i]['context']['contexts']) + " </context>"
    extra_info.append({
        "grounding_question": "Output \"yes\" or \"no\" as your final answer.\n" + context + ds[i]['question'],
        "grounding_answer": ds[i]['final_decision'].capitalize()
    })
    
    if i > 3000:
        break
```

```python
import os

os.makedirs("pubmedqa", exist_ok=True)
```

```python
import pandas as pd

df = pd.DataFrame.from_records({
    "extra_info": extra_info
})

df[:500].to_parquet("pubmedqa/train.parquet")
df[500:600].to_parquet("pubmedqa/valid.parquet")
df[600:1600].to_parquet("pubmedqa/test.parquet")
```
