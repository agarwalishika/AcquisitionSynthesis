````python
from datasets import load_dataset
ds = load_dataset("open-r1/codeforces-cots", "solutions")['train']
````

````python
extra_info = []

replace_str = "Put your final solution within a single code block:\n```cpp\n<your code here>\n```"
for i in range(len(ds)):
    if replace_str in ds[i]['prompt']:
        extra_info.append({
            "grounding_question": ds[i]['prompt'].replace(replace_str, "").strip(),
            "grounding_answer": ds[i]['generation'].split("```cpp")[-1].split("```")[0].strip()
        })
    
    if i > 3000:
        break
````

````python
import os

os.makedirs("/home/ec2-user/grpo_synthesis/data/codeforces", exist_ok=True)
````

````python
import pandas as pd

df = pd.DataFrame.from_records({
    "extra_info": extra_info
})

df[:500].to_parquet("/home/ec2-user/grpo_synthesis/data/codeforces/train.parquet")
df[500:600].to_parquet("/home/ec2-user/grpo_synthesis/data/codeforces/valid.parquet")
df[600:1600].to_parquet("/home/ec2-user/grpo_synthesis/data/codeforces/test.parquet")
````
