```python
from datasets import load_dataset

ds = load_dataset("KbsdJames/Omni-MATH")['test'].select(range(0,1000))
```

```python
ds
```

```python
extra_info = []

for i in range(len(ds)):
    extra_info.append({
        "grounding_question": ds[i]['problem'].strip(),
        "grounding_answer": str(ds[i]['answer'])
    })
```

```python
import os

os.makedirs("omnimath", exist_ok=True)
```

```python
import pandas as pd

df = pd.DataFrame.from_records({
    "extra_info": extra_info
})

df.to_parquet("omnimath/test.parquet")
```
